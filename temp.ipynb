```python
import json
from pathlib import Path

# 使用 Path 处理路径，自动解决 Windows 系统的反斜杠转义问题
source_path = Path(r"D:\desktop\intern\java\sky_take_out\my\sky-take-out\k6_test\summary_k6_1774753914866.json")
target_path = Path(r"D:\desktop\summary_k6_1774753914866.json")

# 1. 读取数据
if source_path.exists():
    with source_path.open('r', encoding='utf-8') as f:
        json_data = json.load(f)

    # 2. 写入数据（注意模式必须为 'w'）
    with target_path.open('w', encoding='utf-8') as f:
        json.dump(json_data, f, indent=4, ensure_ascii=False)
    
    print(f"文件已成功复制并格式化至: {target_path}")
else:
    print("源文件不存在，请检查路径。")
```
