```python
import torch
# 1. 获取显卡信息
fraction = 0.6
num_of_gpu = 4

total_memory = torch.cuda.get_device_properties(num_of_gpu).total_memory
target_memory = int(total_memory * fraction)

# 2. 计算 float32 张量的元素数量 (float32 占用 4 bytes)
# n = target_memory / 4
n_elements = target_memory // 4

print(f"正在尝试占用 GPU {0}:")
print(f"总显存: {total_memory / 1024**2:.2f} MB")
print(f"目标占用 ({fraction*100}%): {target_memory / 1024**2:.2f} MB")

print("申请显存中...")
dummy_tensor = torch.empty((n_elements,), device=f'cuda:{0}', dtype=torch.float32)
```
